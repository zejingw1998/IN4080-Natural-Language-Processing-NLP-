### 1.3 Tokenization and Bag-of-Words with Scikit-Learn

We need numerical features for the text classifier.

The `CountVectorizer` class from scikit-learn converts our text documents into a to convert a **Bag-of-Words (BoW) representation** , where each column corresponds to a word in the vocabulary and each value represents how many times that word occurs in a document. 


<details>
<summary> Example from rewiews to vocabulary to feature vectors </summary>

If we have:

```text
Rewiew 0: "The movie is boring"
Rewiew 1: "The movie is fun"
```

We want a bag of words:

```text
["The", "boring", "fun", "movie", "is"]
```

and we can represent the reviews as word counts (feature vectors):

|              | boring | good | is | movie | the |
|--------------|--------|------|----|-------|------|
| Review 0     | 1      | 0    | 1  | 1     | 1    |
| Review 1     | 0      | 1    | 1  | 1     | 1    |

</details>
 

We will use the method `fit_transform(raw_documents=...)` which first will fit the vectorizer on the training data to learn the vocabulary, and then transform it into vectors.

**important**: We only use this function on the training data, because we only want to learn the vocabulary from the training data. we use `transform()` on the val and test data. 

Learn about `CountVectorizer` ([here](https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html#sklearn.feature_extraction.text.CountVectorizer))



The following commands may help you inspect the features:

```python
len(cv.vocabulary_)        # Number of words in the BOW
cv.vocabulary_["boring"]   # Index of "boring" in the feature matrix
```

Instantiate a vectorizer, then fit it to and transform the training dataset. How many words are in the bag-of-words?

````python
from sklearn.feature_extraction.text import CountVectorizer

#initialize the vectorizer
cv = CountVectorizer()

# fit the vectorizer to the training data
X_train_bow = cv.fit_transform(X_train)


#print the vocabulary
print(cv.vocabulary_)

#print the number of unique words in the BOW
print(len(cv.vocabulary_))

#look up the index of the word "boring"
print(cv.vocabulary_["boring"])
````
